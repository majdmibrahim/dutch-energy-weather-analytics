**General profile — annual file**


Row and column counts, inferred schema, and missing values per column (count and %).

```python

def profile(df, name):
    n_rows = df.count()
    print(f"===== {name} =====")
    print(f"Rows: {n_rows} | Columns: {len(df.columns)}\n")
    df.printSchema()

    null_row = df.select([
        F.sum(F.col(c).isNull().cast("int")).alias(c) for c in df.columns
    ]).collect()[0].asDict()

    nulls = spark.createDataFrame(
        [(c, int(v), round(100 * v / n_rows, 1)) for c, v in null_row.items()],
        ["column", "null_count", "null_pct"]
    ).orderBy(F.desc("null_count"))
    display(nulls)

profile(annual_raw, "ANNUAL (82610ENG)")
```

Where are the annual nulls? (by energy source)


A null is only meaningful once we know *where* it is. For each source: number of nulls and the first year with data, for production (`_2`), installations (`_7`) and capacity (`_8`).

```python
cols = ["GrossProductionOfElectricity_2",
        "InstallationsInstalledEndOfYear_7",
        "ElectricalCapacityEndOfYear_8"]

display(annual_raw.groupBy("EnergySourcesTechniques").agg(
    F.count("*").alias("rows"),
    *[F.sum(F.col(c).isNull().cast("int")).alias(f"null_{c.split('_')[-1]}") for c in cols],
    *[F.min(F.when(F.col(c).isNotNull(), F.col("Periods"))).alias(f"first_year_{c.split('_')[-1]}") for c in cols]
).orderBy("EnergySourcesTechniques"))
```

**Finding:** production nulls come from technologies that did not exist yet (co-firing from 1995, co-digestion from 2005, offshore wind from 2006). Capacity is never reported for the biomass totals or the overall total. The table is **hierarchical**: totals sit next to their components.

 

Check: which year is missing installation counts for hydro and wind?


```python
display(annual_raw.filter(F.col("InstallationsInstalledEndOfYear_7").isNull() &
        F.col("EnergySourcesTechniques").isin("Hydropower", "Onshore wind energy", "Wind energy, total"))
        .select("EnergySourcesTechniques", "Periods"))
```

Duplicate checks — technical key and business key

`ID` is only a row number added by CBS (technical key). The real identity of a row is the **business key**:

- Annual: `EnergySourcesTechniques` + `Periods`
- Monthly: `Periods`
| Case | Meaning |
|---|---|
| B | fully identical rows |
| B' | identical data, different `ID` |
| A | same `ID` repeated |
| A' | same business key with conflicting data |

```python
def check_duplicates(df, name, business_key):
    n = df.count()
    print(f"===== {name} =====")
    print("B  - identical rows (all columns):  ", n - df.dropDuplicates().count())
    print("B' - identical rows (excluding ID): ", n - df.drop("ID").dropDuplicates().count())
    print("A  - repeated ID:                   ", df.groupBy("ID").count().filter("count > 1").count())

    dup_keys = df.groupBy(*business_key).count().filter("count > 1")
    n_dup = dup_keys.count()
    print(f"A' - repeated business key {business_key}: {n_dup}\n")
    if n_dup > 0:
        display(df.join(dup_keys.select(*business_key), business_key).orderBy(*business_key))

check_duplicates(annual_raw,  "ANNUAL",  ["EnergySourcesTechniques", "Periods"])
check_duplicates(monthly_raw, "MONTHLY", ["Periods"])
```

**Finding:** no duplicates of any kind in either file.

Monthly file — period types, coverage, nulls by period type

`Periods` mixes years, quarters and months. Expected counts (computed by hand before running):

| Type | Expected |
|---|---|
| year | 1929–2025 = 97 |
| quarter | 50 × 4 + 2 = 202 |
| month | 50 × 12 + 7 = 607 |

```python
monthly_typed = (monthly_raw
    .withColumn("year", F.substring("Periods", 1, 4).cast("int"))
    .withColumn("period_type",
        F.when(F.col("Periods").rlike("^[0-9]{4}$"), "year")
         .when(F.col("Periods").rlike("(?i)quarter"), "quarter")
         .otherwise("month")))

display(monthly_typed.groupBy("period_type").count())

display(monthly_typed.orderBy("ID").select("ID", "Periods").limit(5))
display(monthly_typed.orderBy(F.desc("ID")).select("ID", "Periods").limit(5))

display(monthly_typed.groupBy("year").count()
                     .filter(F.col("count") != 17).orderBy("year"))

key_cols = ["WindEnergyTotal_12", "SolarPhotovoltaic_15", "Biomass_9", "HydroPower_11"]
display(monthly_typed.groupBy("period_type")
        .agg(*[F.sum(F.col(c).isNull().cast("int")).alias(f"null_{c}") for c in key_cols]))

display(monthly_typed.filter((F.col("period_type") == "month") &
                             F.col("WindEnergyTotal_12").isNotNull())
                     .agg(F.min("year").alias("first_year_with_monthly_wind")))
```

**Finding:** quarters match (202), but 96 year rows were found vs 97 expected, and 608 months vs 607. One annual row was misclassified as a month. Monthly wind data starts in 2002; solar, hydro and biomass in 2015.

which "month" row does not contain a month name?

```python
month_names = "January|February|March|April|May|June|July|August|September|October|November|December"
display(monthly_typed.filter((F.col("period_type") == "month") &
                             ~F.col("Periods").rlike(month_names))
                     .select("ID", "Periods"))
```

Text quality — hidden whitespace in `Periods`

The misclassified row looks like `2021`, yet it fails `^[0-9]{4}$`. Inspect its length and raw bytes.

```python
display(monthly_raw.filter(F.col("ID") == 828).select(
    "Periods",
    F.length("Periods").alias("length"),
    F.hex(F.col("Periods")).alias("hex")
))
```

**Finding:** length 5, hex `3230323120` → a trailing space (`20`). Check every text column in both files.

```python
def check_whitespace(df, name, cols):
    for c in cols:
        n = df.filter(F.col(c) != F.trim(F.col(c))).count()
        print(f"{name}.{c}: {n} values with leading/trailing spaces")

check_whitespace(monthly_raw, "MONTHLY", ["Periods"])
check_whitespace(annual_raw,  "ANNUAL",  ["EnergySourcesTechniques"])
```

**Sources:** CBS StatLine — `82610ENG` (Renewable electricity; production and capacity, annual) and `84575ENG` (Electricity balance sheet; supply and consumption, monthly).

| | Annual (82610ENG) | Monthly (84575ENG) |
|---|---|---|

| Columns | 11 | 32 |

| Structure | long: 15 categories × 36 years (complete balanced panel) | wide: one row per period, mixed year / quarter / month |
| Duplicates (technical key `ID` + business key) | 0 | 0 |
| Nulls | structural, all explained (see below) | structural, all explained (see below) |
| Text quality | no whitespace issues | 6 `Periods` values with trailing spaces |

### Observations
1. **Mixed granularity (monthly):** each year from 1976 has 1 annual + 4 quarterly + 12 monthly rows (17 rows). 1929–1975 have annual rows only; 2026 has 7 months + 2 quarters. Summing the file as-is would triple-count.
2. **Hierarchical categories (annual):** totals and subtotals (*Total renewable energy sources*, *Wind energy, total*, *Total biomass*, *Total biogas*) sit alongside their components. Summing all rows would multiply totals.
3. **Structural nulls:** co-firing of biomass (from 1995), co-digestion of manure (from 2005) and offshore wind (from 2006) did not exist earlier. Interconnector nulls (Norway, UK, Denmark) reflect cables not yet built. GrossProduction is missing for 1929–1935.
4. **Capacity coverage:** no installed capacity is reported for the biomass totals or the overall renewable total, so capacity factor is only possible for wind (onshore/offshore), solar and hydro. Installation counts are sparse (solar only from 2021).
5. **Provisional data:** 2025 figures are provisional; installation counts for hydro and wind are not yet published for 2025.
6. **Monthly coverage by source:** wind is available monthly from 2002 (~290 months); solar, hydro and biomass monthly only from 2015 (~139 months).
7. **Hidden whitespace (systematic):** 6 `Periods` values contain a trailing space: March 1995–1999 (`"1995 March "`, length 11) and the 2021 annual row (`"2021 "`, hex `3230323120`). The pattern suggests a removed footnote marker in the CBS export rather than random entry errors. The 2021 row was misclassified as a month (96 year rows found vs 97 expected); the March rows would silently become null on date parsing. Fix: `trim()` before any parsing or join.

- **Analysis:** use non-normalised production (`_2`) for weather analysis; `_2 − _1` gives CBS's estimate of the weather effect per year.
